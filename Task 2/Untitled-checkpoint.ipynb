# Convert data into proper date format

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].dtype
```

# Convert time

```python
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M').dt.time
```

# Create useful date features

```python
df['Year'] = df['Date'].dt.year
```

```python
df['Month'] = df['Date'].dt.month
```

```python
df['Month_Name'] = df['Date'].dt.month_name()
```

```python
df['Day'] = df['Date'].dt.day
```

```python
df['Day_Name'] = df['Date'].dt.day_name()
```

# Analyze the most important statistics

```python
df[['Unit price', 'Quantity', 'Total', 'cogs', 'gross income', 'Rating']].mean()
```

```python
df[['Unit price', 'Quantity', 'Total', 'cogs', 'gross income', 'Rating']].median()
```

```python
df[['Unit price', 'Quantity', 'Total', 'cogs', 'gross income', 'Rating']].min()
```

# Analyze sales by product line

```python
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

product_sales
```

```python
product_average = df.groupby('Product line')['Total'].mean().sort_values(ascending=False)

product_average
```

# Analyze sales by branch

```python
branch_sales = df.groupby('Branch')['Total'].sum().sort_values(ascending=False)

branch_sales
```

# Analyze sales by city

```python
city_sales = df.groupby('City')['Total'].sum().sort_values(ascending=False)

city_sales
```

# Analyze customer types

```python
customer_sales = df.groupby('Customer type')['Total'].sum()

customer_sales
```

```python
df.groupby('Customer type')['Total'].mean()
```

# Analyze gender

```python
gender_sales = df.groupby('Gender')['Total'].sum()

gender_sales
```

# Analyze payment methods

```python
payment_count = df['Payment'].value_counts()

payment_count
```

# Analyze quantity

```python
quantity_by_product = (
    df.groupby('Product line')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

quantity_by_product
```

# Analyze gross income

```python
income_by_product = (
    df.groupby('Product line')['gross income']
    .sum()
    .sort_values(ascending=False)
)

income_by_product
```

# Analyze daily sales trend

```python
daily_sales = df.groupby('Date')['Total'].sum()

daily_sales.head()
```

# Analyze monthly sales

```python
monthly_sales = (
    df.groupby(df['Date'].dt.to_period('M'))['Total']
    .sum()
)

monthly_sales
```

# Find correlation

```python
correlation = df[
    ['Unit price', 'Quantity', 'Tax 5%', 'Total',
     'cogs', 'gross income', 'Rating']
].corr()

correlation
```

# Create correlation heatmap

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Between Numerical Variables')
plt.show()
```

# Chart 1: Sales by product line

```python
plt.figure(figsize=(10, 6))

product_sales.plot(kind='bar')

plt.title('Total Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Chart 2: Sales by branch

```python
plt.figure(figsize=(7, 5))

branch_sales.plot(kind='bar')

plt.title('Total Sales by Branch')
plt.xlabel('Branch')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

# Chart 3: Daily sales trend

```python
plt.figure(figsize=(12, 5))

daily_sales.plot(kind='line')

plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Chart 4: Sales distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Total'], bins=20)

plt.title('Distribution of Sales')
plt.xlabel('Total Sales')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

# Chart 5: Quantity vs Total sales

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Quantity'], df['Total'])

plt.title('Quantity vs Total Sales')
plt.xlabel('Quantity')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

# Chart 6: Payment method distribution

```python
plt.figure(figsize=(7, 5))

payment_count.plot(kind='bar')

plt.title('Payment Method Usage')
plt.xlabel('Payment Method')
plt.ylabel('Number of Transactions')

plt.tight_layout()
plt.show()
```

# Optional: Customer type comparison

```python
plt.figure(figsize=(7, 5))

customer_sales.plot(kind='bar')

plt.title('Total Sales by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```
